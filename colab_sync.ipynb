# Link a session's video to an audio recording

1. **Runtime → Change runtime type → T4 GPU** (the spoken-word check uses Whisper).
2. In the annotation app: open the **audio** recording → **Copy video sync key** (under the player). The app shows which session zip it found; confirm it.
3. Run the cells below and paste the key when asked.
4. When it says **Done**, load the recording again in the app: the video appears above the audio.

The zip and audio are downloaded to this Colab machine only, and deleted when the script finishes. The camera files are added next to the audio in Box (skipped if they're already there), and a small sync file next to the shared transcripts. If the recording has no shared transcript yet, it's transcribed and shared too, so you don't need to run the transcription notebook for it.
